```python
import pandas as pd

url = "https://raw.githubusercontent.com/sherintht/Retail-Sales-Analysis-using-Python-and-Power-BI/main/retail_sales_dataset.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset size:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Converting the date column into date format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Remove exact duplicate rows
df = df.drop_duplicates().copy()

print("Data cleaning completed!")
print("Dataset shape after cleaning:", df.shape)

display(df.head())
```

```python
print("Total Sales:", df["Total Amount"].sum())

print("Average Sales:", df["Total Amount"].mean())

print("Median Sales:", df["Total Amount"].median())

print("Standard Deviation:", df["Total Amount"].std())

print("Total Quantity Sold:", df["Quantity"].sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

category_sales = df.groupby(
    "Product Category"
)["Total Amount"].sum().sort_values(ascending=False)

print(category_sales)

category_sales.plot(kind="bar", figsize=(9, 5))

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_sales = df.groupby(
    "Month"
)["Total Amount"].sum()

print(monthly_sales)

monthly_sales.plot(
    kind="line",
    marker="o",
    figsize=(12, 5)
)

plt.title("Monthly Retail Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
numeric_data = df.select_dtypes(include="number")

correlation = numeric_data.drop(
    columns=["Transaction ID"],
    errors="ignore"
).corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
